# JSON Data Exploration

```python
import json

try:
    with open('source-data.json', 'r') as file:
        data = json.load(file)
    print("JSON data loaded successfully.")
except FileNotFoundError:
    print("Error: 'source-data.json' not found. Make sure the file is in the same directory as the notebook.")
except json.JSONDecodeError:
    print("Error: Could not decode JSON. Check the file for syntax errors.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
if 'data' in locals():
    if isinstance(data, dict):
        print("Top-level keys:")
        for key in data.keys():
            print(f"- {key}")
    elif isinstance(data, list):
        print(f"JSON is a list with {len(data)} elements. Displaying indices as top-level keys:")
        for i in range(len(data)):
            print(f"- Index {i}")
    else:
        print(f"JSON is a {type(data).__name__}. Cannot display top-level keys in the usual sense.")
else:
    print("Data not loaded. Please run the first cell to load the JSON data.")
```

```python
if 'data' in locals():
    if isinstance(data, dict):
        print("Data types of top-level key-value pairs:")
        for key, value in data.items():
            print(f"- Key: '{key}', Type: {type(value).__name__}")
    elif isinstance(data, list):
        print("Data types of top-level elements:")
        for i, item in enumerate(data):
            print(f"- Index {i}, Type: {type(item).__name__}")
    else:
        print(f"JSON is a {type(data).__name__}. Cannot summarize data types in the usual sense.")
else:
    print("Data not loaded. Please run the first cell to load the JSON data.")
```

```python
def get_element_type_summary(arr):
    if not arr:
        return "empty array"
    types = set(type(elem).__name__ for elem in arr)
    if len(types) == 1:
        return f"array of {types.pop()}s"
    else:
        return f"array of mixed types: {', '.join(sorted(list(types)))}"

def explore_json_element(element, indent=''):
    """Recursively explores a JSON element and prints its structure."""
    if isinstance(element, dict):
        print(f"{indent}Type: dict, Keys: {len(element)}")
        for key, value in element.items():
            print(f"{indent}  Key: '{key}'")
            explore_json_element(value, indent + '    ')
    elif isinstance(element, list):
        type_summary = get_element_type_summary(element)
        print(f"{indent}Type: list, Length: {len(element)}, Contents: {type_summary}")
        if len(element) > 0 and isinstance(element[0], (dict, list)): # Only explore if elements are complex
             # To avoid overly verbose output for lists of simple types, 
             # we can choose to explore only the first element, or a sample.
             # Here, we'll explore each element if it's a dict or list.
            for i, item in enumerate(element):
                print(f"{indent}  Index: {i}")
                explore_json_element(item, indent + '    ')
    elif isinstance(element, str):
        print(f"{indent}Type: str")
    elif isinstance(element, (int, float)):
        print(f"{indent}Type: number ({type(element).__name__})")
    elif isinstance(element, bool):
        print(f"{indent}Type: bool")
    elif element is None:
        print(f"{indent}Type: null")
    else:
        print(f"{indent}Type: {type(element).__name__}")

if 'data' in locals():
    print("Detailed JSON Structure:")
    explore_json_element(data)
else:
    print("Data not loaded. Please run the first cell to load the JSON data.")
```
